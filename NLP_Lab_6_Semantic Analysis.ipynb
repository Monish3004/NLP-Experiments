```python
import spacy
nlp = spacy.load('en_core_web_sm')
sample_text = input("Enter Sentence: ")
doc = nlp(sample_text)
print([(token.text, token.pos_) for token in doc])
```

```python
nlp = spacy.load("en_core_web_sm")
doc = nlp(sample_text)

print("\nDependency Relations:")
for token in doc:
    print(f"Word: {token.text}, Head: {token.head.text}, Dependency: {token.dep_}")
```

```python
# Install necessary libraries
#%pip install language_tool_python

import language_tool_python

# Function to check and correct grammar
def check_grammar(input_text):
    tool = language_tool_python.LanguageTool('en-US')  # English US
    matches = tool.check(input_text)
    
    if matches:
        corrected_text = language_tool_python.utils.correct(input_text, matches)
        print(f"Original Text: {input_text}")
        print(f"Corrected Text: {corrected_text}")
    else:
        print(f"No grammatical errors found in: '{input_text}'")
        
# Sample sentence to check
sample_text = "i am sam and i cook well"  # This is grammatically incorrect
check_grammar(sample_text)
```

```python
from transformers import MarianMTModel, MarianTokenizer

# Function to perform translation
def translate(text, src_lang="en", tgt_lang="fr"):
    model_name = f'Helsinki-NLP/opus-mt-{src_lang}-{tgt_lang}'
    
    # Load the model and tokenizer
    tokenizer = MarianTokenizer.from_pretrained(model_name)
    model = MarianMTModel.from_pretrained(model_name)
    
    # Tokenize the input text
    inputs = tokenizer(text, return_tensors="pt", padding=True, truncation=True)
    
    # Generate translation using the model
    translated = model.generate(**inputs)
    
    # Decode the generated translation
    translated_text = tokenizer.decode(translated[0], skip_special_tokens=True)
    return translated_text

# Example usage
text_to_translate = "Machine translation is an exciting field of NLP."
translated_text = translate(text_to_translate, src_lang="en", tgt_lang="fr")
print(f"Original text: {text_to_translate}")
print(f"Translated text: {translated_text}")
```

```python
from googletrans import Translator

translator = Translator()

text_to_translate = "அய்யா விளையாட்டு மைதானம் வந்தார்."
translated_text = translator.translate(text_to_translate, src="ta", dest="en").text

print(f"Original text: {text_to_translate}")
print(f"Translated text (Tamil to English): {translated_text}")
```
